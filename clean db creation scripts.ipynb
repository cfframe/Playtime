Clean up files (preparatory to comparing elsewhere):
- remove time elements from timestamps
- replace KB/MB/GB etc sizes with dummy text

```python
folder_path = 'C:\\temp\\edm\\database\\'
```

```python
import os
import re
import sys
import time

output_folder_path = os.path.join(folder_path, 'cleaned')

if not os.path.exists(output_folder_path):
    try:
        os.makedirs(output_folder_path)
    except Exception as e:
        print(f"Error creating output folder '{output_folder_path}': {e}")
        sys.exit(1)

```

```python
def load_file(file_path):
    if not os.path.exists(file_path):
        print(f"Error: The file '{file_path}' does not exist.")
        sys.exit(1)

    try:
        with open(file_path, 'r') as file:
            content = file.read()
            return content
    except Exception as e:
        print(f"Error reading the file '{file_path}': {e}")
        sys.exit(1)
```

```python
time_pattern = re.compile(r'\d{2}:\d{2}:\d{2}')
def remove_times(content):
    lines = content.splitlines()
    new_lines = []
    for line in lines:
        # Remove time patterns
        cleaned_line = re.sub(time_pattern, '', line)
        # Strip leading and trailing whitespace
        cleaned_line = cleaned_line.strip()
        if cleaned_line:  # Only add non-empty lines
            new_lines.append(cleaned_line)
    return '\n'.join(new_lines)
```

```python
size_pattern = re.compile(r'\d+(\.\d+)?\s*(KB|MB|GB|TB|PB|EB|ZB|YB)')
def replace_variable_sizes(content):
    lines = content.splitlines()
    new_lines = []
    for line in lines:
        # Replace variable sizes with dummy text
        cleaned_line = re.sub(size_pattern, 'XXX', line)
        new_lines.append(cleaned_line)
    return '\n'.join(new_lines)
```

```python
# Get list of files in the folder
def get_files_in_folder(folder_path):
    try:
        return [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f)) and f.endswith('.txt')]
    except Exception as e:
        print(f"Error accessing folder '{folder_path}': {e}")
        sys.exit(1)

files = get_files_in_folder(folder_path)
```

```python
# Clean all txt files in the folder
def clean_files_in_folder(folder_path):
    files = get_files_in_folder(folder_path)
    for file_name in files:
        file_path = os.path.join(folder_path, file_name)
        content = load_file(file_path)
        cleaned_content = remove_times(content)
        cleaned_content = replace_variable_sizes(cleaned_content)

        output_file_name = file_name.replace('.txt', ' cleaned.txt')
        output_file_path = os.path.join(output_folder_path, output_file_name)

        try:
            with open(output_file_path, 'w') as output_file:
                output_file.write(cleaned_content)
            print(f"Cleaned content saved to '{output_file_path}'") 
        except Exception as e:
            print(f"Error saving the cleaned content: {e}")
            sys.exit(1)

clean_files_in_folder(folder_path)
print(f"All files cleaned successfully. Time: {time.strftime('%Y-%m-%d %H:%M:%S')}")
```
