```python
class Top:
    main = 'Top'
    sub = 'Sub'
class Left(Top):
    sub = 'Left'
class Right(Top) :
    sub = 'Right'

class Bottom(Left, Right):
    pass
c = Bottom()
print(f'Left, Right: c.main: {c.main}, c.sub: {c.sub}')


# Gives TypeError: Cannot create a consistent method resolution order (MRO) for bases Top, Right
# class Bottom(Top, Right):
#     pass
# c = Bottom()
# print(f'c.main: {c.main}, c.sub: {c.sub}')

class Bottom(Left, Top):
    pass
c = Bottom()
print(f'Left, Top: c.main: {c.main}, c.sub: {c.sub}')

class Bottom(Right, Left):
    pass
c = Bottom()
print(f'Right, Left: c.main: {c.main}, c.sub: {c.sub}')

class Bottom(Right):
    pass
c = Bottom()
print(f'Right: c.main: {c.main}, c.sub: {c.sub}')

class Bottom(Right, Left, Top):
    pass
c = Bottom()
print(f'Right, Left, Top: c.main: {c.main}, c.sub: {c.sub}')

for base in Bottom.__bases__:
    print(f'base.__name__: {base.__name__} base.main: {base.main} base.sub: {base.sub}; isinstance(c, base): {isinstance(c, base)}')
```

```python
class Collection:
    stamps = 2

    def __init__(self, stuff):
        self.stuff = stuff

    def dispose(self) :
        del self.stuff

print('--- set binder to Collection(1) ---')
binder = Collection(1)
print('-- Collection.__dict__ and binder.__dict__ before dispose() --')
for e in Collection.__dict__:
    print(f'Collection.{e} = {Collection.__dict__[e]}')
print('-- binder.__dict__ --')
for e in binder.__dict__:
    print(f'binder.{e} = {binder.__dict__[e]}')

binder.dispose()
print('-- Collection.__dict__ and binder.__dict__ after dispose() --')
for e in Collection.__dict__:
    print(f'Collection.{e} = {Collection.__dict__[e]}')
print('-- binder.__dict__ --')
for e in binder.__dict__:
    print(f'binder.{e} = {binder.__dict__[e]}')
```
