```python
class Top:
    main = 'Top'
    sub = 'Sub'
class Left(Top):
    sub = 'Left'
class Right(Top) :
    sub = 'Right'

class Bottom(Left, Right):
    pass
c = Bottom()
print(f'Left, Right: c.main: {c.main}, c.sub: {c.sub}')


# Gives TypeError: Cannot create a consistent method resolution order (MRO) for bases Top, Right
# class Bottom(Top, Right):
#     pass
# c = Bottom()
# print(f'c.main: {c.main}, c.sub: {c.sub}')

class Bottom(Left, Top):
    pass
c = Bottom()
print(f'Left, Top: c.main: {c.main}, c.sub: {c.sub}')

class Bottom(Right, Left):
    pass
c = Bottom()
print(f'Right, Left: c.main: {c.main}, c.sub: {c.sub}')

class Bottom(Right):
    pass
c = Bottom()
print(f'Right: c.main: {c.main}, c.sub: {c.sub}')

class Bottom(Right, Left, Top):
    pass
c = Bottom()
print(f'Right, Left, Top: c.main: {c.main}, c.sub: {c.sub}')

for base in Bottom.__bases__:
    print(f'base.__name__: {base.__name__} base.main: {base.main} base.sub: {base.sub}; isinstance(c, base): {isinstance(c, base)}')
```

```python
class Collection:
    stamps = 2

    def __init__(self, stuff):
        self.stuff = stuff

    def dispose(self) :
        del self.stuff

print('--- set binder to Collection(1) ---')
binder = Collection(1)
print('-- Collection.__dict__ and binder.__dict__ before dispose() --')
for e in Collection.__dict__:
    print(f'Collection.{e} = {Collection.__dict__[e]}')
print('-- binder.__dict__ --')
for e in binder.__dict__:
    print(f'binder.{e} = {binder.__dict__[e]}')

binder.dispose()
print('-- Collection.__dict__ and binder.__dict__ after dispose() --')
for e in Collection.__dict__:
    print(f'Collection.{e} = {Collection.__dict__[e]}')
print('-- binder.__dict__ --')
for e in binder.__dict__:
    print(f'binder.{e} = {binder.__dict__[e]}')
```

Exam block #4: Object oriented programming

```python
# Inheritance - ok
class A:
    def do(self):
        print("A")


class BL(A):
    def do(self):
        print("BLeft")


class BR(A):
    def do(self):
        print("BRight")


class C(BR, BL):
    pass


o = C()
o.do()

# print(issubclass(A, A), issubclass(A, BL)), issubclass(A, BR), issubclass(A, C)
print(issubclass(A, A), issubclass(BL, A), issubclass(BR, A), issubclass(C, A))
print(isinstance(o, A), isinstance(o, BL), isinstance(o, BR), isinstance(o, C))
```

```python
# Inheritance - ok
class A:
    def do(self):
        print("A")

class B(A):
    def do(self):
        print("B")

class C(A):
    def do(self):
        print("C")

class D(B, C):
    pass

o = D()
o.do()
d_bases = [base.__name__ for base in D.__bases__]
print(f'D bases: {d_bases}')
for c in [A, B, C, D]:
    print(f'isinstance(o, {c.__name__}): {isinstance(o, c)}')
```

```python
def double_step(n):
    for i in range(0, 2*n, 2):
        yield i

for v in double_step(5):
    print(v, end=" ")
```

```python
def Fib(n):
    p = pp = 1
    for i in range(n):
        if i in [0, 1]:
            yield 1
        else:
            n = p + pp
            pp, p = p, n
            yield n

print(list(Fib(10)))
y = Fib(6)
for _ in range(6):
    print(next(y))
```

```python
# Binary file streams
byte = bytearray(1)
bytes_95 = bytearray(95)
try:
    out_stream = open("file.bin", "wb")
    for i in range(32,127):
        byte[0] = i;
        out_stream.write(byte)
    out_stream.close();
    in_stream = open("file.bin", "rb")
    # First way to read bytes
    # Bytes are read into variable bytes_95, the number of bytes read is returned by readinto()
    read_in = in_stream.readinto(bytes_95)
    in_stream.close()
    print(read_in,"byte(s) read in")
    for i in range(95):
        print(chr(bytes_95[i]), end='')
except IOError as exception:
    print("I/O error occurred:", os.strerror(exception.errno))
print()
print('*','*','*','*')
```
